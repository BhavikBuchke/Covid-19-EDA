```python
# Downloads the COVID-19 clinical trials dataset using KaggleHub and retrieves the local file path for further processing
# Dataset Download

import kagglehub

# Download latest version

path = kagglehub.dataset_download("parulpandey/covid19-clinical-trials-dataset")
print("Path to dataset files:", path)
```

```python
# Load Dataset

import pandas as pd
import os

# Reads the CSV file into a Pandas DataFrame and displays the first few records to inspect structure and columns.

df = pd.read_csv(os.path.join(path, 'COVID clinical trials.csv'))
df.head(3)
```

```python
# Tail Inspection
# Displays the last few rows to verify data completeness and detect formatting anomalies.

df.tail()
```

```python
# Data Overview
# Uses df.info() to examine data types & non-null counts.

df.info()
```

```python
# Drop Unnecessary Columns
# Removes irrelevant columns such as funding details and URLs that are not required for analysis.

drop_col = ['Funded Bys', 'Other IDs', 'URL',]
df = df.drop(columns=drop_col, errors='ignore')
```

```python
# Missing Value Analysis
# Calculates total missing values and identifies columns containing null entries.

print(f'Total empty values in dataset: {df.isnull().sum().sum()}')
print(f'Columns with empty values:\n{df.isnull().sum()[df.isnull().sum()>0]}')
```

```python
# Drop Sparse Rows
# Removes rows with excessive missing values using a threshold to retain meaningful records.

df = df.dropna(thresh=len(df.columns) - 5)
df.head(3)
```

```python
# Drop High-Missing Columns
# Removes columns like Study Documents and Results First Posted due to high missing rates.

df = df.drop(columns=['Study Documents', 'Results First Posted', 'Acronym'])
```

```python
# Fill Categorical Missing Values
# Replaces missing categorical values (Phases, Gender, Locations) with "Unknown".s

categorical_columns = ['Phases', 'Gender', 'Locations']
for col in categorical_columns:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')
```

```python
# Fill Numerical Missing Values
# Fills missing Enrollment values using median imputation to reduce skew impact.

df['Enrollment'] = df['Enrollment'].fillna(df['Enrollment'].median())
```

```python
# Duplicate Check
# Identifies duplicate rows to ensure dataset integrity.

duplicate_rows = df.duplicated().sum()
if duplicate_rows > 0:
  print(f'Sum of values in duplicate rows:{duplicate_rows}')
else:
  print('No duplicate rows found.')
```

```python
# Date Conversion
# Converts date columns to datetime format to enable time-series and duration analysis.

date_cols = ['Start Date', 'Primary Completion Date',
             'Completion Date', 'First Posted', 'Last Update Posted']
for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], format='mixed')
        print(df[col].dtype)
```

```python
# Text Cleaning
# Standardizes text fields by stripping whitespace and converting to title case.

text_cols = ['Phases', 'Status', 'Gender', 'Age']
for col in text_cols:
    if col in df.columns:
        df[col] = df[col].str.strip().str.title()

df[text_cols].head(3)
```

```python
# Saving processed data as csv file.
df.to_csv('processed_data.csv', index=False)
```

```python
# Summary Statistics
# Generates descriptive statistics for numerical columns.

print(f'Summary Statistics:\n {df.describe()}')
```

```python
# Enrollment Distribution
# Visualizes enrollment size using log scale due to high variance between small and large trials.

import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
# Use log scale to handle the massive difference between small and large trials
sns.histplot(df['Enrollment'], bins=50, log_scale=True, color='teal')
plt.title('Distribution of Trial Enrollment Sizes (Log Scale)', fontsize=16)
plt.xlabel('Number of Participants (Log Scale)')
plt.show()
```

### **Conclusion:**
The enrollment distribution is highly right-skewed, indicating that the majority of clinical trials involve small participant groups, while a limited number of large-scale trials account for very high enrollments. This reflects the typical structure of clinical research where early-stage or pilot studies dominate in count, and large Phase 3 trials are comparatively fewer but resource-intensive.

```python
# Monthly Trial Trend
# Resamples data monthly to observe how clinical trial launches evolved over time.

sns.set_style("darkgrid")
plt.rcParams['figure.figsize'] = (12, 6)
monthly_trials = df.set_index('Start Date').resample('ME').size()

plt.figure()
sns.lineplot(data=monthly_trials, marker='o', color='green', linewidth=2.5)
plt.title('Timeline of COVID-19 Clinical Trials (Monthly)', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Number of New Trials', fontsize=12)
plt.show()
```

### **Conclusion:**
The time-series trend reveals a sharp surge in clinical trial initiations during the early stages of the pandemic, followed by gradual stabilization. This pattern reflects the urgent global research response to COVID-19 and the subsequent normalization as treatment protocols and vaccines became established.

```python
# Country Extraction
# Extracts country names from the Locations column for geographic analysis.

df.loc[:, 'Country'] = df['Locations'].str.split(',').str[-1].str.strip()
```

```python
#Geographic Analysis
# Displays the top 10 countries conducting COVID-19 clinical trials.

top_countries = df['Country'].value_counts().head(10)

plt.figure()
sns.barplot(x=top_countries.values, y=top_countries.index, hue=top_countries.index, palette='viridis', legend=False)
plt.title('Top 10 Countries Conducting Clinical Trials', fontsize=16)
plt.xlabel('Number of Trials', fontsize=12)
plt.show()
```

### **Conclusion:**
Clinical research activity is concentrated in a small number of countries, indicating disparities in research infrastructure, funding capacity, and healthcare systems. Developed nations dominate trial participation, highlighting global inequality in research leadership and resource allocation.

```python
# Trial Status Distribution
# Shows the distribution of trial statuses (e.g., Recruiting, Completed).

plt.figure()
status_counts = df['Status'].value_counts().head(5)
sns.barplot(x=status_counts.values, y=status_counts.index, hue=status_counts.index, palette='coolwarm', legend=False)
plt.title('Current Status of Clinical Trials', fontsize=16)
plt.xlabel('Count', fontsize=12)
plt.show()
```

### **Conclusion:**
A significant portion of trials are either completed or actively recruiting, demonstrating rapid mobilization of global research efforts. However, the presence of terminated or withdrawn trials suggests operational challenges, funding limitations, or evolving scientific understanding during the pandemic.

```python
# Duration Calculation
# Calculates trial duration in days and removes negative durations caused by data inconsistencies.
df['Duration_Days'] = (df['Completion Date'] - df['Start Date']).dt.days

# Remove negative durations (data entry errors)
df = df[df['Duration_Days'] > 0]
```

```python
# Duration Distribution
# Plots distribution of clinical trial durations.

plt.figure()
sns.histplot(df['Duration_Days'], bins=30, kde=True, color='purple')
plt.title('Distribution of Clinical Trial Durations (Days)', fontsize=16)
plt.xlabel('Duration (Days)', fontsize=12)
plt.xlim(0, 1500) # Limit x-axis to remove extreme outliers for better visibility
plt.show()
```

### **Conclusion:**
Trial durations vary substantially, though most studies fall within a moderate time range. Extremely short durations may indicate early-phase or observational studies, while longer durations likely correspond to advanced-phase trials requiring extended monitoring for safety and efficacy.

```python
# Intervention Type Extraction
# Extracts high-level intervention types (Drug, Device, Biological, etc.) from text.

df['Intervention_Type'] = df['Interventions'].astype(str).apply(lambda x: x.split(':')[0] if ':' in x else 'Other')
```

```python
# Top Intervention Types
# Visualizes most common intervention categories.
top_interventions = df['Intervention_Type'].value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_interventions.values, y=top_interventions.index, hue=top_interventions.index, palette='magma', legend=False)
plt.title('Top Intervention Types in COVID-19 Trials', fontsize=16)
plt.xlabel('Count')
plt.show()
```

### **Conclusion:**
Drug-based interventions represent the largest category, reflecting the immediate need for therapeutic solutions during the pandemic. Biological interventions, devices, and other modalities appear less frequently, indicating prioritization of pharmacological research over alternative strategies.

```python
# Phase Simplification Function
# Defines a function to categorize complex trial phase descriptions into simplified stage groups.

def simplify_phase(phase):
    phase = str(phase)
    if 'Phase 4' in phase: return 'Phase 4 (Post-Market)'
    elif 'Phase 3' in phase: return 'Phase 3 (Late Stage)'
    elif 'Phase 2' in phase: return 'Phase 2 (Mid Stage)'
    elif 'Phase 1' in phase: return 'Phase 1 (Early Stage)'
    elif 'Not Applicable' in phase: return 'Not Applicable'
    else: return 'Other'

df['Phase_Category'] = df['Phases'].apply(simplify_phase)
```

```python
# Phase Distribution Plot
# Displays distribution of simplified clinical trial stages.

plt.figure(figsize=(10, 6))
sns.countplot(y='Phase_Category', data=df, order=df['Phase_Category'].value_counts().index, hue='Phase_Category', palette='Set2', legend=False)
plt.title('Distribution of Clinical Trial Stages', fontsize=16)
plt.show()
```

### **Conclusion:**
The majority of trials fall within mid-to-late-stage phases (Phase 2 and Phase 3), suggesting a strong focus on testing treatment effectiveness and safety at scale. Early-stage trials are fewer, and post-market (Phase 4) studies are limited, which aligns with the timeline of rapid vaccine and drug development.
